## Programação Orientada a Objetos

```python
class CountingClicker:
    """ A classe pode/deve ter um docstring, como as funções"""
    def __init__(self, count = 0):
        self.count = count

    def __repr__(self):
        return f"CountingClicker(count = {self.count})"

    def click(self, num_times = 1):
        """Clique no contador algumas vezes"""
        self.count += num_times

    def read(self):
        return self.count

    def reset(self):
        self.count = 0 
```

```python
clicker1 = CountingClicker()
print(f"Nenuhm clique: {clicker1.count}")

clicker2 = CountingClicker(100)
print(f"Cem cliques: {clicker2.count}")
```

```python
clicker = CountingClicker()
```

```python
assert clicker.read() == 0, "Clicker should start with count 0"
```

```python
clicker.click()
clicker.click()
```

```python
assert clicker.read() == 2, "after two clicks, clocker should have count 2"
```

```python
clicker.reset()
```

```python
assert clicker.read() == 0, "after reset, clicker should be back to 0"
```

```python
class NoResetClicker(CountingClicker):
    # Está subclasse tem os mesmos métodos da CoutingClicker
    # Mas seu método 
    def reset(self):
        pass
```

```python
clicado = NoResetClicker()
```

```python
assert clicado.read() == 0
```

```python
clicado.click()
```

```python
assert clicado.read() == 1
```

```python
clicado.reset()
```

```python
assert clicado.read() == 1, "reset shouldn't do anything"
```

### args e kwargs

```python
def doubler(f): # 
    def g(x): # quando g é chamado com um argumento ele já tem a função f(f1) de doubler
        return 2 * f(x) # aqui o f é a função f1, que soma mais 1 a x e depois multiplica por 2
    return g
```

```python
def f1(x):
    return x + 1
```

```python
g = doubler(f1) # a função f1 é passada como argumento 
print(g)
```

```python
assert g(3) == 8
```

```python
assert g(-1) == 0
```

```python
def f2(x,y):
    return x + y
```

```python
g = doubler(f2)
try:
    g(1,2)
except TypeError:
    print("Como definido, a função g só recebe um argumento")
```

```python
def magic(*args, **kwargs): 
    print("unnamed args", args) # args é uma tupla para argumentos sem nome
    print("keyword args:", kwargs) # kwargs é um dicionário para argumentos com nome

magic(1,2, key="word", key2="word2")
```

```python
def other_way_magic(x,y,z):
    return x + y + z

x_y_list = [1, 2]
z_dict = {"z": 3}
assert other_way_magic(*x_y_list, **z_dict) == 6
```

```python
def twice(repeater: Callable[[str, int], str], s:str) -> str:
    return repeater(s,5)

def comma_repeater(s: str, n: int) -> str:
    n_copies = [s for _ in range(n)]
    return ', '.join(n_copies)
```
