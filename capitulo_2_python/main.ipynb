## Dicionários

```python
empty_dict = {} # forma pythonica
empty_dict2 = dict() # outra forma de inicializar, podendo usar argumentos
notas = {"Joel": 80, "Tim": 95} # forma literal
print(empty_dict)
print(empty_dict2)
print(notas)

joels_grade = notas["Joel"]
print(f"Nota do Joel: {joels_grade}")
```

```python
# Se pesquisar por uma chave que não está no dicionário dará erro
try:
    kates_grade = notas["Kate"]
except KeyError:
    print("Não tem nota para Kate!")
```

```python
# Para evitar o erro usar o operador "in"
joels_has_grade = "Joel" in notas
print(joels_has_grade)
kate_has_grade = "Kate" in notas
print(kate_has_grade)
```

```python
# O método "get" retorna um valor padrão no lugar de um exceção
joels_grade = notas.get("Joel", 0) 
print(joels_grade)
kates_grade = notas.get("Kate", "Não tem") # O segundo argumento é o valor padrão retornado caso não ache a chave desejada
print(kates_grade)
no_ones_grade = notas.get("Ninguém") # Também pode passar só o primeiro argumento da chave que está procurando, se não existir retorna None como valor padrão
print(no_ones_grade)
```

```python
notas["Tim"] = 99 # substituindo o valor anterior
notas["Kate"] = 100 # Adiciona um novo part de chave e valor
num_students = len(notas)
print(num_students)
print(notas)
```

```python
s = set() # forma correta de criar um conjunto vazio
print(f"Conjunto s: {s}")
s.add(1)
print(f"Conjunto s após adicionar 1: {s}")
s.add(2)
print(f"Conjunto s após adicionar 2: {s}")
s.add(2)
print(f"Conjunto s após adicionar 2 novamente: {s}") # Só pode ter elementos únicos

print(f"Tamanho do conjunto s: {len(s)}")
x = 2 in s
print(f"2 está no conjunto s? {x}")
y = 3 in s
print(f"3 está no conjunto s? {y}")
```

## Compreensões de Listas

```python
even_numbers = [x for x in range(5) if x % 2 == 0]
print(f"Lista de números pares: {even_numbers}")

squares = [x * x for x in range(5)]
print(f"Lista de quadrados: {squares}")

even_squares = [x * x for x in even_numbers]
print(f"Lista de quadrados de números pares: {even_squares}")
```

```python
# Dá para transformas listas em conjuntos e dicionários usando compreensões de listas
square_dict = {x: x * x for x in range(5)}
print(f"Dicionário de números quadrados: {square_dict}")

square_set = {x * x for x in [1, -1]}
print(f"Conjunto de quadrados: {square_set}")
```

```python
zeros = [0 for _ in even_numbers]
print(f"Lista de zeros: {zeros}")
```

```python
pairs = [(x,y)
         for x in range(10)
         for y in range(10)]
print(f"Lista de pares: {pairs}")
```

```python
increasing_pairs = [(x,y)
                    for x in range(10)
                    for y in range(x+1,10)]
print(f"Lista de pares: {increasing_pairs}")
```
