# Capítulo 3 - Visualizando Dados

```python
from matplotlib import pyplot as plt
```

```python
anos = [1950, 1960, 1970, 1980, 1990, 2000, 2010]
gdp = [300.2, 543.3, 1075.9, 2862.5, 5979.6, 10289.7, 14958.3]
```

```python
plt.plot(anos, gdp, color = 'green', marker = 'o', linestyle = 'solid')
plt.title("Nominal GDP")
plt.ylabel("Bilhões de $")
plt.show()
```

## Gráfico de Barras

```python
movies = ["Annie Hall", "Ben-Hur", "Casablanca", "Gandhi", "West Side Story"]
num_oscars = [5, 11, 3, 8, 10]
```

```python
plt.bar(range(len(movies)), num_oscars)
plt.title("Quantidade de Oscars de cada filme")
plt.ylabel("# Premios da academia")
plt.xticks(range(len(movies)), movies)
plt.show()
```

```python
from collections import Counter
```

```python
grades = [83, 95, 91, 87, 70, 0, 85, 82, 100, 67, 73, 77, 0]
```

```python
histogram = Counter(min(grades // 10 * 10, 90) for grades in grades)
print(histogram)
```

```python
plt.bar([x + 5 for x in histogram.keys()],
    histogram.values(),
    10,
    edgecolor = (0, 0, 0))

plt.axis([-5, 105, 0, 5])

plt.xticks([10 * i for i in range(11)])
plt.xlabel("Decile")
plt.ylabel("# of Students")
plt.title("Distribution of Exam 1 grades")
plt.show()
```

```python
mentions = [500, 505]
years = [2017, 2018]
```

```python
plt.bar(years, mentions, 0.8)
plt.xticks(years)
plt.ylabel("# N° de vezes que ouvi alguém dizer 'Data Science'")
plt.ticklabel_format(useOffset=False)
plt.axis([2016.5, 2018.5, 499, 506])
plt.title("Observe esse aumento 'Imenso'!")
plt.show()
```

```python
plt.bar(years, mentions, 0.8)
plt.xticks(years)
plt.ylabel("# N° de vezes que ouvi alguém dizer 'Data Science'")
plt.ticklabel_format(useOffset=False)
plt.axis([2016.5, 2018.5, 0, 550])
plt.title("O aumento diminuiu bastante")
plt.show()
```
