```python
# Função que soma 2 vetores de tamanhos iguais
def add(v: Vector, w: Vector) -> Vector:
    assert len(v) == len(w), "Os vetores devem ter o mesmo tamanho!"
    return [v_i + w_i for v_i, w_i in zip(v,w)]
```

```python
assert add([1,2], [4,5,6]) == [5,7,9]
```

```python
soma = add([1,2,3], [4,5,6])
print(soma)
```

```python
# Função que subtrai 2 vetores de tamanhos iguais
def subtract(v: Vector, w: Vector) -> Vector:
    assert len(v) == len(w), "Os vetores devem ter o mesmo tamanho!"
    return [v_i - w_i for v_i, w_i in zip(v,w)]
```

```python
assert subtract([5,7,9], [4,5,6]) == [1,2,3]
```

```python
subtracao = subtract([5,7,9], [4,5,6])
print(subtracao)
```

```python
# Soma de multiplus vetores
def vector_sum(vectors: List[Vector]) -> Vector:
    assert vectors, "Nenhum vetor!!!"

    num_elements = len(vectors[0])
    assert all(len(v) == num_elements for v in vectors), "Vetores de tamanhos diferentes"

    return [sum(vector[i] for vector in vectors) 
            for i in range(num_elements)]
```

```python
assert vector_sum([[1,2], [3,4], [5,6], [7,8]]) == [16, 20]
```

```python
soma_vetores = vector_sum([[1,2], [3,4], [5,6], [7,8]])
print(soma_vetores)
```

```python
# Multiplicação de um vetor por um escalar
def scalar_multiply(c: float, v: Vector) -> Vector:
    return [c * v_i for v_i in v]
```

```python
assert scalar_multiply(2, [1,2,3]) == [2,4,6]
```

```python
vetor_escalar = scalar_multiply(2, [1,2,3])
print(f"Multiplicação de um vetor por um escalar: {vetor_escalar}")
```

```python
# Média de vários vetores
def vector_mean(vectors: List[Vector]) -> Vector:
    n = len(vectors)
    return scalar_multiply(1/n, vector_sum(vectors))
```

```python
assert vector_mean([[1,2], [3,4], [5,6]]) == [3,4]
```

```python
media_vetores = vector_mean([[1,2], [3,4], [5,6]])
print(f"Média dos vetores: {media_vetores}")
```

```python
# Produto escalar
def dot(v: Vector, w: Vector) -> float:
    assert len(v) == len(w), "Os vetores devem ter o mesmo tamanho"

    return sum(v_i * w_i for v_i, w_i in zip(v,w))
```

```python
assert dot([1,2,3],[4,5,6]) == 32 # 1*4 + 2*5 + 3*6
```

```python
prod_escalar = dot([1,2,3],[4,5,6])
print(f"O produto escalar entre os vetores é: {prod_escalar}")
```

```python
# Soma dos quadrados de um vetor
def sum_of_squares(v: Vector) -> float:
    return dot(v,v)
```

```python
assert sum_of_squares([1,2,3]) == 14
```

```python
sum_squares = sum_of_squares([1,2,3])
print(f"Soma dos quadrados de um vetor: {sum_squares}")
```

```python
import math
```

```python
# Magnitude é o comprimento do vetor
def magnitude(v: Vector) -> float:
    return math.sqrt(sum_of_squares(v)) # raiz quadrada da soma dos quadrados
```

```python
assert magnitude([3,4]) == 5
```

```python
mag = magnitude([3,4])
print(f"A magnitude do vetor [3,4] é: {mag}")
```

```python
# Computar a distância entre dois vetores
def squared_distance(v: Vector, w: Vector) -> float:
    return sum_of_squares(subtract(v,w))
```

```python
def distance(v: Vector, w: Vector) -> float:
    return math.sqrt(squared_distance(v,w))
```

```python
distancia = distance([1,2], [3,4])
print(f"A distância entre os vetores [1,2] e [3,4] é: {distancia}")
```

```python
# Equivalente
def distance2(v: Vector, w: Vector) -> float:
    return magnitude(subtract(v, w))
```

```python
dist = distancia = distance2([1,2], [3,4])
print(f"A distância entre os vetores [1,2] e [3,4] é: {dist}")
```

```python
from typing import Tuple
```

```python
# retorna o número de linhas e colunas da matriz
def shape(A: Matrix) -> Tuple[int, int]:
    num_rows = len(A) # 
    num_cols = len(A[0]) if A else 0
    return num_rows, num_cols
```

```python
assert shape([[1,2,3], [4,5,6]]) == (2,3)
```

```python
tam_matriz = shape([[1,2,3], [4,5,6]])
print(f"O tamanho da matriz é: {tam_matriz}")
```

```python
def get_row(A: Matrix, i: int) -> Vector:
    return A[i]
```

```python
def get_column(A: Matrix, j: int) -> Vector:
    return [A_i[j]
                for A_i in A]
```

```python
from typing import Callable
```

```python
def make_matrix(num_rows: int,
                num_cols: int,
                entry_fn: Callable[[int, int], float]) -> Matrix:
    return [[entry_fn(i,j)
             for j in range(num_cols)]
             for i in range(num_rows)]
```

```python
def identity_matrix(n: int) -> Matrix:
    return make_matrix(n,n, lambda i, j: 1 if i == j else 0)
```

```python
matriz_identidade = identity_matrix(10)

print(f"Criando uma matriz identidade de tamanho {len(matriz_identidade)}: ")
for i in range(len(matriz_identidade)):
    print(matriz_identidade[i])
```
