```python
from collections import Counter
import matplotlib.pyplot as plt
from typing import List
```

```python
num_friends = [100,49,41,40,25,23,15,17,13,22,67,50,50,49,23,25,25,15,13,13,13]
```

```python
friends_counts = Counter(num_friends)
xs = range(101)
ys = [friends_counts[x] for x in xs]
plt.bar(xs,ys)
plt.axis([0, 101, 0, 25])
plt.title("Histograma das Contagens de Amigos")
plt.xlabel("nº de amigos")
plt.ylabel("nº de pessoas")
plt.show()
```

```python
num_points = len(num_friends)
print(f"Número de pontos de dados: {num_points}")
```

```python
largest_value = max(num_friends)
smallest_value = min(num_friends)

print(f"Maior número de amigos: {largest_value}")
print(f"Menor número de amigos: {smallest_value}")
```

```python
sorted_values = sorted(num_friends)
print(f"Lista de amigos ordenada: {sorted_values}")
```

```python
smallest_value = sorted_values[0]
print(f"Menor valor da lista ordenada: {smallest_value}")
```

```python
second_smallest_value = sorted_values[1]
print(f"Segundo menor valor da lista ordenada: {second_smallest_value}")
```

```python
second_largest_value = sorted_values[-2]
print(f"Segundo maior valor da lista ordenada: {second_largest_value}")
```

## Tendências centrais

```python
def mean(xs: List[float]) -> float:
    return sum(xs) / len(xs)

mean(num_friends)
```

```python
def _median_odd(xs: List[float]) -> float:
    return sorted(xs)[len(xs) // 2]
```

```python
def _median_even(xs: List[float]) -> float:
    sorted_xs = sorted(xs)
    hi_midpoint = len(xs) // 2
    return (sorted_xs[hi_midpoint - 1] + sorted_xs[hi_midpoint ]) / 2
```

```python
def median(v: List[float]) -> float:
    return _median_even(v) if len(v) % 2 == 0 else _median_odd(v)
```

```python
print(f"Mediana do número de amigos: {median(num_friends)}")
```
